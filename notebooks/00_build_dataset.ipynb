# Refrigerant Viscosity Dataset Builder

This notebook builds a reproducible dataset of **refrigerant dynamic viscosities** as functions of temperature and pressure. It also associates each fluid with molecular-graph descriptors derived from its SMILES representation. The final artifact, `data/fluids_data.pkl`, is consumed by the downstream graph neural network (GNN) training workflow.

## Pipeline overview

1. **Scrape the NIST WebBook fluid index** to obtain the list of fluids it provides thermophysical data for.
2. **Cross-reference with CoolProp / pyfluids** to build a unified refrigerant table, filtering to pure halocarbons and classifying each entry (CFC / HCFC / HFC / HFO / HCFO / PFC).
3. **Query viscosity** on a regular `(T, P)` grid, preferring CoolProp data and falling back to the NIST WebBook where CoolProp is unavailable.
4. **Convert SMILES to molecular graphs** (PyTorch Geometric `Data` objects) using the helper in `src/fluid_processing.py`.
5. **Save** the slim long-form table to disk.
6. **Visualise** isotherms of viscosity vs. pressure, grouped by refrigerant class.

## External dependencies

- `requests`, `urllib3` &mdash; HTTP access to the NIST WebBook. TLS verification is disabled in the current implementation to accommodate corporate proxy interception; use normal certificate verification whenever the network environment permits it.
- `pyfluids`, `CoolProp` &mdash; reference equations of state and SMILES/formula lookup.
- `pandas`, `numpy`, `tqdm`, `matplotlib` &mdash; data wrangling and plotting.
- Local `src/fluid_processing.py` &mdash; provides `smiles_to_graph`.

> **Execution note:** The WebBook-query and graph-construction stages depend on external services and computational chemistry tooling, so they can take several minutes. Run the cells in order and inspect the reported coverage before overwriting the persisted dataset.


## 1. Scrape the NIST WebBook fluid index

The NIST WebBook fluid landing page (<https://webbook.nist.gov/chemistry/fluid/>) contains a `<select>` element listing every fluid for which an equation of state is available. Each `<option>` carries:

- a `value` attribute &mdash; the **WebBook ID** (e.g. `C811972`) used in subsequent data-table URLs.
- the option text &mdash; a human readable name, often including the ASHRAE R-number in parentheses (e.g. `1,1,1,2-Tetrafluoroethane (R134a)`).

We extract these pairs with a regular expression and load them into a `fluids` DataFrame. TLS verification is disabled because corporate proxies frequently re-sign HTTPS traffic, breaking certificate validation for this public read-only page.


```python
import re
import urllib3
import requests
import pandas as pd

URL = "https://webbook.nist.gov/chemistry/fluid/"

# Corporate proxy may rewrite the TLS chain; disable verification for this public page.
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)
html = requests.get(URL, timeout=30, verify=False).text

# Extract all <option value="..."> ... </option> entries from the species select
pattern = re.compile(r'<option[^>]*value="([^"]+)"[^>]*>([^<]+)</option>', re.IGNORECASE)
options = [(v, n.strip()) for v, n in pattern.findall(html)]

fluids = pd.DataFrame(options, columns=["id", "name"])
print(f"Total fluids: {len(fluids)}")
fluids
```

## 2. Build the unified refrigerant table

Two complementary sources of refrigerant data are merged here:

| Source | Strength | Limitation |
|--------|----------|------------|
| **NIST WebBook** | High-fidelity multi-parameter equations of state, recognised reference data. | Naming via ASHRAE R-numbers requires regex extraction; coverage focused on common refrigerants. |
| **CoolProp** (via `pyfluids.FluidsList`) | Programmatic access; provides molecular **formula** and **SMILES** strings. | Stereochemistry written with parentheses (`R1234ze(E)`) instead of bare suffix (`R1234zeE`). |

### Key steps

- **Name harmonisation** &mdash; `to_coolprop_name` rewrites `R1234zeE` &rarr; `R1234ze(E)` so the same R-number maps to either source.
- **R-number extraction** from WebBook names uses the pattern `(R<digits><suffix>)`.
- **Pure-fluid filter** for CoolProp: rejects blends (single trailing uppercase letter such as `R407C`), explicit `*Mix` names and the entire 500-series (azeotropes).
- **Formula parsing** &mdash; CoolProp returns LaTeX-flavoured formulae like `C_{3}Cl_{1}F_{3}H_{2}`. The regex `([A-Z][a-z]?)_\{(\d+)\}` produces an `{element: count}` dict.
- **Double-bond detection** &mdash; an `=` in the SMILES indicates an olefinic C=C bond; for entries with a missing/`?` SMILES we fall back to the ASHRAE convention that 4-digit numbers starting with `1` (e.g. `R1234yf`) are unsaturated.

### Classification rules

| Class | C | H | Cl | F | C=C |
|------|---|---|----|----|-----|
| **PFC**  | yes | no  | no  | yes | no  |
| **CFC**  | yes | no  | yes | yes | no  |
| **HCFC** | yes | yes | yes | yes | no  |
| **HFC**  | yes | yes | no  | yes | no  |
| **HFO**  | yes | yes | no  | yes | yes |
| **HCFO** | yes | yes | yes | yes | yes |
| **Other**| anything outside `{C, H, Cl, F, O}` (e.g. iodine in `R13I1`) |

The cell prints the per-class and per-source counts and returns the `refrigerants` DataFrame for downstream use.


```python
# Build a unified refrigerant table from two sources:
#   - NIST WebBook  (fluids DataFrame above)
#   - CoolProp      (pyfluids.FluidsList)
#
# Element counts and SMILES (used to detect a C=C double bond) are queried
# directly from CoolProp via get_fluid_param_string -> no hard-coded tables.
#
# Each row is classified as CFC / HCFC / HFC / HFO / HCFO / PFC
# (Other otherwise), based on elemental composition and the presence of a
# C=C double bond.
#
# Classification rules:
#   PFC  : C, F                 (no H, no Cl)
#   CFC  : C, Cl, F             (no H, no C=C)
#   HCFC : C, H, Cl, F          (no C=C)
#   HFC  : C, H, F              (no Cl, no C=C)
#   HFO  : C, H, F + C=C        (no Cl)
#   HCFO : C, H, Cl, F + C=C

from pyfluids import FluidsList
from CoolProp.CoolProp import get_fluid_param_string


# --- Map our R-number key to the CoolProp name ---------------------------
# CoolProp uses parenthesised stereo descriptors (e.g. "R1234ze(E)") whereas
# pyfluids/our table uses "R1234zeE".
_stereo_re = re.compile(r"^(R\d+[a-z]+)([EZ])$")

def to_coolprop_name(r_number: str) -> str:
    m = _stereo_re.match(r_number)
    return f"{m.group(1)}({m.group(2)})" if m else r_number


def cp_param(r_number: str, param: str):
    """Fetch a CoolProp fluid parameter, trying both naming conventions."""
    for name in (to_coolprop_name(r_number), r_number):
        try:
            return get_fluid_param_string(name, param)
        except Exception:
            continue
    return None


# --- Parse CoolProp's LaTeX-style formula, e.g. "C_{3}Cl_{1}F_{3}H_{2}" --
_element_re = re.compile(r"([A-Z][a-z]?)_\{(\d+)\}")

def parse_formula(formula_str):
    if not formula_str:
        return {}
    return {el: int(n) for el, n in _element_re.findall(formula_str)}


def classify(elements: dict, has_double_bond: bool) -> str:
    has_C  = elements.get("C", 0)  > 0
    has_H  = elements.get("H", 0)  > 0
    has_Cl = elements.get("Cl", 0) > 0
    has_F  = elements.get("F", 0)  > 0
    # Anything beyond C/H/Cl/F/O (e.g. iodine in R13I1) -> Other
    extras = set(elements) - {"C", "H", "Cl", "F", "O"}
    if extras or not has_C or not has_F:
        return "Other"
    if has_double_bond:
        return "HCFO" if has_Cl else "HFO"
    if not has_H and not has_Cl:
        return "PFC"
    if not has_H and has_Cl:
        return "CFC"
    if has_H and has_Cl:
        return "HCFC"
    return "HFC"  # has_H and not has_Cl


# --- 1. Refrigerants from the NIST WebBook --------------------------------
r_number_re = re.compile(r"\(R(C?\d+[a-zA-Z]*)\)")

def webbook_r_number(name: str):
    m = r_number_re.search(name)
    return f"R{m.group(1)}" if m else None

webbook_df = (
    fluids.assign(r_number=fluids["name"].map(webbook_r_number))
          .dropna(subset=["r_number"])
          .rename(columns={"id": "webbook_id", "name": "webbook_name"})
          [["r_number", "webbook_name", "webbook_id"]]
          .reset_index(drop=True)
)

# --- 2. Refrigerants from CoolProp / pyfluids -----------------------------
# A pyfluids name is a *pure* refrigerant if it starts with 'R' followed by
# digits and the character after the digits is NOT a single uppercase letter
# (uppercase = blend designator, e.g. R407C, R410A) and the name does not
# end with 'Mix'.
pure_re = re.compile(r"^R(?:C)?\d+[A-Za-z0-9]*$")

def is_pure_pyfluid(name: str) -> bool:
    if not pure_re.match(name) or name.endswith("Mix"):
        return False
    if re.fullmatch(r"R5\d{2}", name):  # 500-series are azeotropic blends
        return False
    suffix = re.sub(r"^R(?:C)?\d+", "", name)
    if suffix == "":
        return True
    if re.fullmatch(r"[A-Z]", suffix):  # single uppercase letter = blend
        return False
    return True

coolprop_names = sorted({f.name for f in FluidsList if is_pure_pyfluid(f.name)})
coolprop_df = pd.DataFrame({"r_number": coolprop_names, "coolprop_name": coolprop_names})

# --- 3. Merge sources -----------------------------------------------------
merged = pd.merge(webbook_df, coolprop_df, on="r_number", how="outer")

def source_label(row):
    in_wb = pd.notna(row["webbook_id"])
    in_cp = pd.notna(row["coolprop_name"])
    if in_wb and in_cp:
        return "both"
    if in_wb:
        return "webbook"
    return "coolprop"

merged["source"] = merged.apply(source_label, axis=1)

# --- 4. Pull formula + SMILES from CoolProp, derive class ----------------
merged["formula_str"] = merged["r_number"].map(lambda r: cp_param(r, "formula"))
merged["smiles"]      = merged["r_number"].map(lambda r: cp_param(r, "SMILES"))

merged["elements"] = merged["formula_str"].map(parse_formula)
for el in ["C", "H", "Cl", "F", "O"]:
    merged[el] = merged["elements"].map(lambda d, el=el: d.get(el, 0) if d else None)

# Detect a C=C double bond from the SMILES string (ignores aromatic notation,
# which doesn't apply to the refrigerants in this list). For entries whose
# CoolProp SMILES is missing/unknown ('?'), fall back to ASHRAE numbering:
# 4-digit R-numbers starting with '1' designate unsaturated halocarbons.
_unsat_re = re.compile(r"^R1\d{3}[a-zA-Z]*$")

def has_double_bond(row):
    smi = row["smiles"]
    if smi and "=" in smi:
        return True
    return bool(_unsat_re.match(row["r_number"]))

merged["double_bond"] = merged.apply(has_double_bond, axis=1)

merged["class"] = merged.apply(
    lambda row: classify(row["elements"], row["double_bond"]) if row["elements"] else "Unknown",
    axis=1,
)

unknown = merged.loc[merged["class"] == "Unknown", "r_number"].tolist()
if unknown:
    print(f"WARNING: no formula retrieved from CoolProp for: {unknown}")

refrigerants = (
    merged[["r_number", "class", "source", "smiles", "C", "H", "Cl", "F", "O",
            "webbook_name", "webbook_id", "coolprop_name"]]
    .sort_values(["class", "r_number"])
    .reset_index(drop=True)
)

print("\nClass counts:")
print(refrigerants["class"].value_counts().to_string())
print("\nSource counts:")
print(refrigerants["source"].value_counts().to_string())
refrigerants
```

## 3. Sample viscosity on a (T, P) grid

We evaluate dynamic viscosity $\mu(T, P)$ on a regular grid:

- **Temperatures**: 10, 30, 50, 70, 90 &deg;C (converted to Kelvin).
- **Pressures**: 10, 20, 30, 40, 50 MPa.

### Two data backends

1. **NIST WebBook *IsoBar* endpoint** &mdash; one request per `(refrigerant, pressure)` returns viscosity at *all* temperatures simultaneously, which is far cheaper than per-point queries. The response is a TSV table parsed with `pandas.read_csv`. Returned values are in &micro;Pa&middot;s and converted to Pa&middot;s. Each table row is matched to the closest requested temperature within a 0.5 K tolerance.
2. **CoolProp `PropsSI("V", "T", T, "P", P, name)`** &mdash; queried per `(T, P)` point, with the same name-fallback logic as before.

### Combining the sources (per fluid)

Source selection is decided **per refrigerant**, not per `(T, P)` point, so isotherm curves never mix backends:

- If **CoolProp** returns a viscosity at *every* `(T, P)` grid point for the fluid &rarr; use CoolProp for that fluid.
- Otherwise, if **NIST WebBook** covers the full grid &rarr; fall back to NIST for that whole fluid.
- If neither source covers the full grid &rarr; the fluid is dropped.

The chosen backend is recorded in the `viscosity_source` column (`"coolprop"` or `"nist"`).

A `tqdm` progress bar tracks the outer `refrigerant &times; pressure` loop. The final `viscosity_data` DataFrame is in **long form** &mdash; one row per `(refrigerant, T, P)`.


```python
# Build a long-form (refrigerant, temperature, pressure, viscosity) table.
#
# For every refrigerant in `refrigerants`, viscosity is evaluated on a grid
# of temperatures and pressures.
#   - NIST WebBook is queried once per (refrigerant, pressure) using the
#     IsoBar endpoint, returning viscosities at all temperatures at once.
#   - CoolProp is queried per (T, P) point for refrigerants that have one.
# Source selection is *per fluid*: CoolProp is used when it covers the
# entire (T, P) grid for that fluid; otherwise the whole fluid falls back
# to NIST WebBook data (used only if NIST itself covers the full grid).

import io
import numpy as np
from tqdm import tqdm
from CoolProp.CoolProp import PropsSI

temperatures_C = np.arange(10, 101, 20)               # 10, 30, 50, 70, 90 C
temperatures   = [t + 273.15 for t in temperatures_C] # K
pressures      = np.arange(10e6, 50e6 + 1, 10e6)      # 10 .. 50 MPa

# --- NIST WebBook isobar query (all temperatures at one pressure) --------
NIST_ISOBAR_URL = (
    "https://webbook.nist.gov/cgi/fluid.cgi?Action=Data&Wide=on"
    "&ID={wid}&Type=IsoBar&Digits=5"
    "&TLow={tlow}&THigh={thigh}&TInc={tinc}&P={p}"
    "&RefState=DEF&TUnit=K&PUnit=MPa&DUnit=mol%2Fl&HUnit=kJ%2Fmol"
    "&WUnit=m%2Fs&VisUnit=uPa*s&STUnit=N%2Fm"
)

def nist_viscosity_isobar(webbook_id, t_values_K, p_Pa):
    """Return {T: viscosity_Pa_s} for the requested temperatures, or {}."""
    if not webbook_id or pd.isna(webbook_id):
        return {}
    tlow, thigh = float(min(t_values_K)), float(max(t_values_K))
    tinc = (thigh - tlow) / (len(t_values_K) - 1) if len(t_values_K) > 1 else 0
    url = NIST_ISOBAR_URL.format(
        wid=webbook_id, tlow=tlow, thigh=thigh, tinc=tinc, p=p_Pa / 1e6
    )
    try:
        text = requests.get(url, timeout=30, verify=False).text
        table = pd.read_csv(io.StringIO(text), sep="\t")
    except Exception:
        return {}
    t_col = next((c for c in table.columns if c.startswith("Temperature")), None)
    v_col = next((c for c in table.columns if c.startswith("Viscosity")),  None)
    if not t_col or not v_col:
        return {}
    out = {}
    for _, row in table.iterrows():
        try:
            T_val = float(row[t_col])
            v_val = float(row[v_col]) * 1e-6   # uPa*s -> Pa.s
        except (TypeError, ValueError):
            continue
        # Match to the closest requested temperature (within 0.5 K)
        nearest = min(t_values_K, key=lambda t: abs(t - T_val))
        if abs(nearest - T_val) < 0.5:
            out[nearest] = v_val
    return out


# --- CoolProp viscosity per point ----------------------------------------
def coolprop_viscosity(r_number, T, P):
    for name in (to_coolprop_name(r_number), r_number):
        try:
            return PropsSI("V", "T", T, "P", P, name)
        except Exception:
            continue
    return None


# --- Build the long-form table -------------------------------------------
records = []
total_iters = len(refrigerants) * len(pressures)
with tqdm(total=total_iters, desc="Refrigerant x Pressure") as pbar:
    for _, ref in refrigerants.iterrows():
        for P in pressures:
            nist_map = nist_viscosity_isobar(ref["webbook_id"], temperatures, P)
            for T in temperatures:
                v_nist = nist_map.get(T)
                v_cp   = coolprop_viscosity(ref["r_number"], T, P)
                records.append({
                    "r_number":           ref["r_number"],
                    "class":              ref["class"],
                    "source":             ref["source"],
                    "smiles":             ref["smiles"],
                    "C":   ref["C"],  "H":  ref["H"],
                    "Cl":  ref["Cl"], "F":  ref["F"], "O": ref["O"],
                    "webbook_name":       ref["webbook_name"],
                    "webbook_id":         ref["webbook_id"],
                    "coolprop_name":      ref["coolprop_name"],
                    "temperature":        T,
                    "pressure":           P,
                    "viscosity_nist":     v_nist,
                    "viscosity_coolprop": v_cp,
                })
            pbar.update(1)

viscosity_data = pd.DataFrame.from_records(records)

# --- Per-fluid source selection ------------------------------------------
# For each refrigerant, prefer CoolProp if it covers the entire (T, P) grid.
# Otherwise, fall back to NIST WebBook for that whole fluid (only if NIST
# itself covers the full grid). Fluids covered by neither source are dropped.
expected = len(temperatures) * len(pressures)

cp_complete = (
    viscosity_data.groupby("r_number")["viscosity_coolprop"]
                  .apply(lambda s: s.notna().sum() == expected)
)
nist_complete = (
    viscosity_data.groupby("r_number")["viscosity_nist"]
                  .apply(lambda s: s.notna().sum() == expected)
)

def pick_source(r):
    if cp_complete.get(r, False):
        return "coolprop"
    if nist_complete.get(r, False):
        return "nist"
    return None

viscosity_data["viscosity_source"] = viscosity_data["r_number"].map(pick_source)

viscosity_data["viscosity"] = np.where(
    viscosity_data["viscosity_source"] == "coolprop",
    viscosity_data["viscosity_coolprop"],
    np.where(
        viscosity_data["viscosity_source"] == "nist",
        viscosity_data["viscosity_nist"],
        np.nan,
    ),
)
# Convert Pa.s -> uPa.s for storage and downstream use.
viscosity_data["viscosity"] = viscosity_data["viscosity"] * 1e6

# Drop fluids with no complete-grid source.
incomplete = sorted(
    r for r in viscosity_data["r_number"].unique() if pick_source(r) is None
)
viscosity_data = (
    viscosity_data.dropna(subset=["viscosity"]).reset_index(drop=True)
)
print(f"Dropped {len(incomplete)} refrigerants with no complete source: {incomplete}")

print("Source counts (per row):")
print(viscosity_data["viscosity_source"].value_counts(dropna=False).to_string())
print("\nFluids per source:")
print(
    viscosity_data.groupby("viscosity_source")["r_number"]
                  .nunique().to_string()
)
print(f"\nTotal rows: {len(viscosity_data)}  "
      f"(refrigerants: {viscosity_data['r_number'].nunique()})")

viscosity_data
```

## 4. Build molecular graphs from SMILES

Each refrigerant is converted into a **PyTorch Geometric** `Data` object using `smiles_to_graph` from the local `src/fluid_processing.py`. Because the molecular structure is independent of `(T, P)`, we:

1. Iterate over the **unique** `(r_number, smiles)` pairs only &mdash; one graph per molecule.
2. **Broadcast** the resulting graphs onto every row of `viscosity_data` via a simple `map`.

### Robustness

- The `src/` directory is added to `sys.path` if not already present.
- `fluid_processing` is **reloaded** if it was imported earlier in the kernel, so edits to the helper file take effect without restarting.
- Refrigerants with a missing/`?` SMILES are recorded in `failed` rather than crashing the loop; any other exception is caught with its message truncated for readability.
- A safety check raises `RuntimeError` if **no** graphs were built &mdash; this avoids silently overwriting `viscosity_data["graph"]` with NaNs and dropping the entire dataset.
- Rows whose refrigerant did not yield a graph are dropped at the end.


```python
# Build a molecular graph per refrigerant from its SMILES.
# Graphs are independent of (T, P) so we compute one per unique r_number
# and broadcast onto every row of `viscosity_data`.

import sys
import importlib
from pathlib import Path

# Make src/fluid_processing.py importable.
src_dir = Path.cwd() / "src"
if src_dir.is_dir() and str(src_dir) not in sys.path:
    sys.path.insert(0, str(src_dir))

# Force a fresh import in case fluid_processing.py changed since first import.
if "fluid_processing" in sys.modules:
    importlib.reload(sys.modules["fluid_processing"])
from fluid_processing import smiles_to_graph

unique_smiles = (
    viscosity_data[["r_number", "smiles"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

graphs = {}
failed = []
for _, row in tqdm(list(unique_smiles.iterrows()), desc="Building graphs"):
    r_number, smi = row["r_number"], row["smiles"]
    if not smi or smi == "?":
        failed.append((r_number, "missing SMILES"))
        continue
    try:
        graphs[r_number] = smiles_to_graph(smi, verbose=False)
    except Exception as e:
        failed.append((r_number, str(e)[:120]))

print(f"Built graphs for {len(graphs)} / {len(unique_smiles)} refrigerants")
if failed:
    print("Failed:")
    for r, msg in failed:
        print(f"  {r}: {msg}")

if not graphs:
    raise RuntimeError(
        "No graphs were built; refusing to wipe viscosity_data. "
        "Fix the underlying error (see 'Failed' above) and re-run this cell."
    )

# Attach to the long-form table (same Data object reused across all (T, P) rows).
viscosity_data["graph"] = viscosity_data["r_number"].map(graphs)

missing = viscosity_data["graph"].isna().sum()
if missing:
    print(f"Dropping {missing} rows whose refrigerant has no graph.")
    viscosity_data = viscosity_data.dropna(subset=["graph"]).reset_index(drop=True)

print(f"\nTotal rows: {len(viscosity_data)}  "
      f"(refrigerants: {viscosity_data['r_number'].nunique()})")
viscosity_data.head()
```

## 5. Persist the slim dataset

We project `viscosity_data` to the **columns required by training** and rename `r_number` to `Name` for compatibility with downstream code. The schema saved to `data/fluids_data.pkl` is:

| Column        | Type             | Meaning                              |
|---------------|------------------|--------------------------------------|
| `Name`        | `str`            | ASHRAE R-number (e.g. `R134a`)       |
| `smiles`      | `str`            | Canonical SMILES from CoolProp       |
| `pressure`    | `float` (Pa)     | State-point pressure                 |
| `temperature` | `float` (K)      | State-point temperature              |
| `viscosity`   | `float` (&micro;Pa&middot;s) | CoolProp viscosity, NIST fallback   |
| `class`       | `str`            | CFC / HCFC / HFC / HFO / HCFO / PFC  |
| `graph`       | `torch_geometric.data.Data` | Molecular graph              |

Pickle is used because the `graph` column holds non-tabular Python objects.


```python
# Slim sub-dataset and save it to pickle.
df = (
    viscosity_data
    .rename(columns={"r_number": "Name"})
    [["Name", "smiles", "pressure", "temperature", "viscosity", "class", "graph"]]
    .reset_index(drop=True)
)

Path("data").mkdir(exist_ok=True)
df.to_pickle("data/fluids_data.pkl")
print(f"Saved {len(df)} rows to data/fluids_data.pkl")
df.head()
```

## 6. Sanity-check plots: viscosity vs. pressure

For each refrigerant **class**, a figure is produced with one **subplot per refrigerant** in that class. Each subplot shows isotherms of viscosity against pressure &mdash; one curve per temperature in the grid.

These plots are diagnostic:

- Smooth, monotonic isotherms suggest both data sources are self-consistent.
- Crossings, kinks or unphysical jumps usually indicate a phase boundary or a mismatch between NIST and CoolProp values for that refrigerant.

Empty subplot slots (when the class size is not a multiple of `n_cols = 4`) are hidden via `axis("off")`. After the figures, a textual listing enumerates the members of each class.


```python
import matplotlib.pyplot as plt

# One figure per refrigerant class, with one subplot per refrigerant.
classes = sorted(viscosity_data["class"].unique())
n_cols = 4

for cls in classes:
    cls_data = viscosity_data[viscosity_data["class"] == cls]
    unique_fluids = sorted(cls_data["r_number"].unique())
    n_fluids = len(unique_fluids)
    n_rows = int(np.ceil(n_fluids / n_cols))

    fig, axes = plt.subplots(n_rows, n_cols,
                             figsize=(20, 5 * n_rows),
                             squeeze=False)
    axes = axes.flatten()
    fig.suptitle(f"Class: {cls}  ({n_fluids} refrigerants)",
                 fontsize=14, fontweight="bold", y=1.00)

    for idx, r_number in enumerate(unique_fluids):
        ax = axes[idx]
        fluid_data = cls_data[cls_data["r_number"] == r_number]

        for temp in sorted(fluid_data["temperature"].unique()):
            temp_data = fluid_data[fluid_data["temperature"] == temp].sort_values("pressure")
            ax.plot(temp_data["pressure"], temp_data["viscosity"],
                    marker="o", markersize=3, linewidth=1.5,
                    label=f"{temp:.1f} K")

        ax.set_xlabel("Pressure (Pa)", fontsize=9)
        ax.set_ylabel("Viscosity (ÂµPaÂ·s)", fontsize=9)
        ax.set_title(f"{r_number} ({cls})", fontsize=10, fontweight="bold")
        ax.legend(title="Temp", fontsize=7, title_fontsize=8, loc="best")
        ax.grid(True, alpha=0.3)
        ax.tick_params(labelsize=8)

    for idx in range(n_fluids, len(axes)):
        axes[idx].axis("off")

    plt.tight_layout()
    plt.show()

print("All plots generated, one figure per class!\n")

# Listing of refrigerants in each class.
print("Refrigerants per class:")
for cls in classes:
    members = sorted(viscosity_data.loc[viscosity_data["class"] == cls, "r_number"].unique())
    print(f"  {cls} ({len(members)}): {', '.join(members)}")
```
