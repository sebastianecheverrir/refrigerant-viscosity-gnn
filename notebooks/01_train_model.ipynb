It loads the graph dataset produced by `00_build_dataset.ipynb`, removes unusable samples and constant graph descriptors, constructs a refrigerant-level train/validation split, normalizes all features using **training data only**, and saves model checkpoints and metric histories under `training_runs/`.

## 2. Load and validate the dataset

Load `data/fluids_data.pkl`, retain refrigerant records, verify the required schema, coerce the thermodynamic variables to numeric values, and remove incomplete rows. The following checks inspect the viscosity range and graph structure before any model inputs are created.


```python
df = pd.read_pickle('data/fluids_data.pkl')

# Keep only refrigerants (names starting with 'R')
df = df[df['Name'].str.startswith('R', na=False)].reset_index(drop=True)

# Validate expected schema for the new supervised format
required_cols = ['graph', 'pressure', 'temperature', 'viscosity']
missing_cols = [c for c in required_cols if c not in df.columns]
if missing_cols:
    raise KeyError(f"Missing required columns in dataframe: {missing_cols}")

# Ensure numeric dtypes for scalar inputs/target
df['pressure'] = pd.to_numeric(df['pressure'], errors='coerce')
df['temperature'] = pd.to_numeric(df['temperature'], errors='coerce')
df['viscosity'] = pd.to_numeric(df['viscosity'], errors='coerce')

df = df.dropna(subset=['graph', 'pressure', 'temperature', 'viscosity']).reset_index(drop=True)

# df['pressure'] = df['pressure']/1e6

df["Name"].unique()
```

```python
# Analyze and drop graph.u attributes that are all zeros across all rows
import numpy as np

# Get the number of u attributes from the first graph
n_attributes = len(df.iloc[0]['graph'].u[0])
print(f"Total number of global attributes (u): {n_attributes}")

# Check which attributes are all zeros across all graphs
all_zero_attributes = []
for attr_idx in range(n_attributes):
    is_all_zero = True
    for idx, row in df.iterrows():
        graph = row['graph']
        # Check if this attribute has any non-zero value across all nodes
        attr_values = [graph.u[node_idx][attr_idx] for node_idx in range(len(graph.u))]
        if any(val != 0 for val in attr_values):
            is_all_zero = False
            break
    if is_all_zero:
        all_zero_attributes.append(attr_idx)

print(f"Attributes with all zeros: {all_zero_attributes}")
print(f"Number of attributes to drop: {len(all_zero_attributes)}")

# Create a set for faster lookup and determine which indices to keep
all_zero_set = set(all_zero_attributes)
indices_to_keep = [i for i in range(n_attributes) if i not in all_zero_set]

# Drop the all-zero attributes from all graphs
if all_zero_attributes and len(indices_to_keep) < n_attributes:
    for idx in range(len(df)):
        graph = df.iloc[idx]['graph']
        # Check if this graph has already been processed
        current_attr_count = graph.u.shape[1] if len(graph.u.shape) == 2 else len(graph.u[0])
        if current_attr_count == n_attributes:
            # Use tensor indexing to keep only non-zero attributes (maintains tensor structure)
            graph.u = graph.u[:, indices_to_keep]
            
            # Also update global_feature_names to match
            if hasattr(graph, 'global_feature_names') and graph.global_feature_names is not None:
                graph.global_feature_names = [graph.global_feature_names[i] for i in indices_to_keep]
    
    print(f"Dropped {len(all_zero_attributes)} attributes. New attribute count: {len(indices_to_keep)}")
elif not all_zero_attributes:
    print("No all-zero attributes found.")
else:
    print("Graphs already processed - skipping modification.")
```

## 3. Construct supervised graph samples

Clone each molecular graph so the persisted dataset remains unchanged, append pressure and temperature to its graph-level feature vector, and store viscosity as the scalar regression target. Every row is therefore one molecular graph evaluated at one thermodynamic state point.


```python
# Prepare the data for training
data_list = []
for _, row in df.iterrows():
    graph = row['graph'].clone()

    # Preserve existing global features and add pressure and temperature
    pressure_temp = torch.tensor([float(row['pressure']), float(row['temperature'])], dtype=torch.float).view(1, -1)
    
    if hasattr(graph, 'u') and graph.u is not None:
        # Concatenate new features to existing global features
        graph.u = torch.cat([graph.u, pressure_temp], dim=1)
    else:
        # No existing global features, just use pressure and temperature
        graph.u = pressure_temp

    # Scalar target: viscosity
    graph.y = torch.tensor([float(row['viscosity'])], dtype=torch.float).view(1, 1)
    data_list.append(graph)

print(f"Total samples created: {len(data_list)}")
print(f"Node features per graph: {data_list[0].x.shape}")
print(f"Global features per graph: {data_list[0].u.shape}")


```

## 4. Split by refrigerant and normalize features

Validation fluids are held out by refrigerant identity rather than by state point. This stricter split measures transfer to unseen molecular graphs and prevents pressureâ€“temperature samples from the same molecule appearing in both subsets. Minâ€“max parameters are fitted exclusively on the training subset and then applied unchanged to validation data. The viscosity target is scaled from a fixed physical lower bound of zero to the maximum training viscosity.


```python

#   CFC (5): R11, R113, R115, R12, R13
#   HCFC (6): R123, R124, R141b, R142b, R21, R22
#   HFC (11): R125, R134a, R143a, R152a, R227ea, R23, R236ea, R236fa, R245fa, R32, R41
#   HFO (2): R1234yf, R1234zeE
#   PFC (4): R116, R14, R218, RC318
#   HCFO (1): R1233zdE (no validation refrigerant here)
```

```python

# Train/validation split by refrigerant name (not by individual samples)
# This ensures refrigerants in training set don't appear in validation set
val_refrigerants = [
    'R12',
    'R124',
    'R125',
    'R1234zeE',
    'R218',
]
train_refrigerants = [r for r in df['Name'].unique() if r not in val_refrigerants]

# Get indices for training and validation based on refrigerant assignment
train_indices = df[df['Name'].isin(train_refrigerants)].index.to_numpy()
val_indices = df[df['Name'].isin(val_refrigerants)].index.to_numpy()

train_data = [data_list[i] for i in train_indices]
val_data = [data_list[i] for i in val_indices]

# Safety check: no overlap between train and validation refrigerants
overlap_refrigerants = set(train_refrigerants).intersection(set(val_refrigerants))
if overlap_refrigerants:
    raise RuntimeError(f"Train/validation refrigerant overlap detected: {sorted(overlap_refrigerants)}")

# Verify no sample index overlap
overlap_indices = set(train_indices).intersection(set(val_indices))
if overlap_indices:
    raise RuntimeError(f"Train/validation sample overlap detected for indices: {sorted(overlap_indices)}")

# Normalize features using min-max scaling based on training data
print("\nNormalizing features using min-max scaling...")

# 1. Normalize node features
train_node_features = torch.cat([data.x for data in train_data], dim=0)
node_min = train_node_features.min(dim=0, keepdim=True).values
node_max = train_node_features.max(dim=0, keepdim=True).values
node_range = node_max - node_min
node_range = torch.where(node_range > 1e-8, node_range, torch.ones_like(node_range))

print(f"Node features - min: {node_min.squeeze()}, max: {node_max.squeeze()}")

for data in train_data:
    data.x = (data.x - node_min) / node_range
for data in val_data:
    data.x = (data.x - node_min) / node_range

print("âœ“ Node features normalized to [0, 1]")

# 2. Normalize edge features (if they exist)
if hasattr(train_data[0], 'edge_attr') and train_data[0].edge_attr is not None:
    train_edge_features = torch.cat([data.edge_attr for data in train_data if data.edge_attr is not None], dim=0)
    edge_min = train_edge_features.min(dim=0, keepdim=True).values
    edge_max = train_edge_features.max(dim=0, keepdim=True).values
    edge_range = edge_max - edge_min
    edge_range = torch.where(edge_range > 1e-8, edge_range, torch.ones_like(edge_range))
    
    print(f"Edge features - min: {edge_min.squeeze()}, max: {edge_max.squeeze()}")
    
    for data in train_data:
        if data.edge_attr is not None:
            data.edge_attr = (data.edge_attr - edge_min) / edge_range
    for data in val_data:
        if data.edge_attr is not None:
            data.edge_attr = (data.edge_attr - edge_min) / edge_range
    
    print("âœ“ Edge features normalized to [0, 1]")
else:
    print("âœ— No edge features to normalize")

# 3. Normalize global features (pressure and temperature)
train_global_features = torch.cat([data.u for data in train_data], dim=0)
global_min = train_global_features.min(dim=0, keepdim=True).values
global_max = train_global_features.max(dim=0, keepdim=True).values
global_range = global_max - global_min
global_range = torch.where(global_range > 1e-8, global_range, torch.ones_like(global_range))

print(f"Global features (P, T) - min: {global_min.squeeze()}, max: {global_max.squeeze()}")

for data in train_data:
    data.u = (data.u - global_min) / global_range
for data in val_data:
    data.u = (data.u - global_min) / global_range

print("âœ“ Global features normalized to [0, 1]")

# 4. Normalize output values (viscosity)
# Use 0 as minimum since viscosity is always positive
y_min = torch.tensor(0.0)
train_output_values = torch.cat([data.y for data in train_data], dim=0)
y_max = train_output_values.max()
y_range = y_max - y_min  # This equals y_max since y_min = 0
if y_range < 1e-8:
    y_range = torch.tensor(1.0)

print(f"Output (viscosity) - min: {y_min.item():.6e} (fixed), max: {y_max.item():.6e} (from training data)")

for data in train_data:
    data.y = (data.y - y_min) / y_range
for data in val_data:
    data.y = (data.y - y_min) / y_range

print("âœ“ Output values normalized to [0, 1]")
print(f"\nNormalization parameters saved for denormalization:")
print(f"  y_min = {y_min.item():.6e}, y_max = {y_max.item():.6e}, y_range = {y_range.item():.6e}")

batch_size = 32
train_loader = DataLoader(train_data, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_data, batch_size=batch_size, shuffle=False)

print(f"\nTraining refrigerants: {len(train_refrigerants)} ({sorted(train_refrigerants)})")
print(f"Validation refrigerants: {len(val_refrigerants)} ({sorted(val_refrigerants)})")
print(f"Training samples: {len(train_data)}")
print(f"Validation samples: {len(val_data)}")
print(f"Input node features: {data_list[0].x.shape[1]}")
print(f"Global features (including pressure & temperature): {data_list[0].u.shape[1]}")
print(f"Output dimension (viscosity): {data_list[0].y.shape[1]}")
```

```python
# Define the Graph Neural Network
class GNN(torch.nn.Module):
    def __init__(self, node_features, hidden_dim, global_features=2):
        super(GNN, self).__init__()

        # Graph attention layers (multi-head with concat=True: output size = hidden_dim * heads)
        self.conv1 = GATConv(node_features, hidden_dim, heads=4, concat=True)
        self.conv2 = GATConv(hidden_dim * 4, hidden_dim, heads=4, concat=True)
        self.conv3 = GATConv(hidden_dim * 4, hidden_dim, heads=4, concat=True)

        # Process global features (pressure/temperature)
        self.global_fc = torch.nn.Linear(global_features, hidden_dim)

        # Final prediction layers (input: hidden_dim*4 from conv3 + hidden_dim from global_fc = hidden_dim*5)
        self.fc1 = torch.nn.Linear(hidden_dim * 4 + hidden_dim, hidden_dim)
        self.fc2 = torch.nn.Linear(hidden_dim, 1)

        self.dropout = torch.nn.Dropout(0.2)

    def forward(self, x, edge_index, batch, u):
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = self.dropout(x)

        x = self.conv2(x, edge_index)
        x = F.relu(x)
        x = self.dropout(x)

        x = self.conv3(x, edge_index)
        x = F.relu(x)

        # Graph-level embedding
        x = global_mean_pool(x, batch)

        # Global features (pressure/temperature) embedding
        u = F.relu(self.global_fc(u))

        # Fusion + scalar viscosity output
        x = torch.cat([x, u], dim=1)
        x = F.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        
        # Ensure positive output (viscosity is always positive)
        x = F.softplus(x)

        return x

# Get dimensions
node_features = data_list[0].x.shape[1]
global_features = data_list[0].u.shape[1]

# Initialize model
hidden_dim = 128
model = GNN(node_features, hidden_dim, global_features)
print(model)
print(f"\nTotal parameters: {sum(p.numel() for p in model.parameters())}")
```

## 5. Define and train the graph attention model

Three multi-head graph-attention layers encode atom-level structure, global mean pooling produces one molecular embedding per graph, and a separate linear layer embeds the global descriptors. The two embeddings are concatenated to predict one positive viscosity value through a softplus output.

Select the optimization objective with `criterion_name`. Training records loss, MAE, and MAPE; improved validation-MAPE and average-MAPE checkpoints are retained without overwriting earlier improvements. Because the configured run uses 3,000 epochs, verify the selected device and output directory before starting this section.


```python
# Training configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)
print(f"Training on: {device}")

# Training function
def train():
    model.train()
    total_loss = 0
    total_mae = 0
    total_mape = 0

    for data in train_loader:
        data = data.to(device)
        optimizer.zero_grad()

        out = model(data.x, data.edge_index, data.batch, data.u)
        loss = criterion(out, data.y)
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * data.num_graphs
        total_mae += torch.sum(torch.abs(out - data.y)).item()
        # Calculate percentage error in normalized space
        total_mape += torch.sum(torch.abs((out - data.y) / (data.y + 1e-8))).item()

    # Descale MAE to original viscosity scale, calculate percentage error
    mape = (total_mape / len(train_loader.dataset)) * 100
    return total_loss / len(train_loader.dataset), (total_mae / len(train_loader.dataset)) * y_range.item(), mape

# Evaluation function
def test(loader):
    model.eval()
    total_loss = 0
    total_mae = 0
    total_mape = 0

    with torch.no_grad():
        for data in loader:
            data = data.to(device)
            out = model(data.x, data.edge_index, data.batch, data.u)
            loss = criterion(out, data.y)
            total_loss += loss.item() * data.num_graphs
            total_mae += torch.sum(torch.abs(out - data.y)).item()
            # Calculate percentage error in normalized space
            total_mape += torch.sum(torch.abs((out - data.y) / (data.y + 1e-8))).item()

    # Descale MAE to original viscosity scale, calculate percentage error
    mape = (total_mape / len(loader.dataset)) * 100
    return total_loss / len(loader.dataset), (total_mae / len(loader.dataset)) * y_range.item(), mape
```

```python
# Get min and max values of y in data_list
all_y = torch.cat([data.y for data in data_list], dim=0)
y_min_all = all_y.min().item()
y_max_all = all_y.max().item()

print(f"Min y in data_list: {y_min_all:.6e}")
print(f"Max y in data_list: {y_max_all:.6e}")
```

```python
# Training loop
from datetime import datetime

# ---------------------------------------------------------------------------
# Select the loss criterion by name. Available options:
#   "mse"      -> standard mean squared error
#   "msre"     -> mean squared relative error
#   "max_msre" -> maximum squared relative error per batch (paper default)
#   "mape"     -> mean absolute percentage error (in fraction, not %)
# ---------------------------------------------------------------------------
criterion_name = "max_msre"

available_criterions = {
    "mse":      torch.nn.MSELoss(),
    "msre":     lambda pred, target: torch.mean(((pred - target) / (target + 1e-18)) ** 2),
    "max_msre": lambda pred, target: torch.max(((pred - target) / (target + 1e-18)) ** 2),
    "mape":     lambda pred, target: torch.mean(torch.abs((pred - target) / (target + 1e-18))),
}

if criterion_name not in available_criterions:
    raise ValueError(
        f"Unknown criterion '{criterion_name}'. "
        f"Choose one of: {list(available_criterions.keys())}"
    )

criterion = available_criterions[criterion_name]
print(f"Using criterion: {criterion_name}")

num_epochs = 3000
train_losses = []
val_losses = []
train_maes = []
val_maes = []
train_mapes = []
val_mapes = []
best_avg_mape = float('inf')
best_val_loss = float('inf')
best_val_mape = float('inf')

# Create a unique output folder for this training run.
# The folder name is prefixed by the chosen criterion so runs are easy to identify.
run_timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
run_output_dir = Path("training_runs") / f"{criterion_name}_{run_timestamp}"
run_output_dir.mkdir(parents=True, exist_ok=True)
print(f"Saving checkpoints to: {run_output_dir}")

# Path to the MAPE history file (rewritten after every epoch)
mape_history_path = run_output_dir / "mape_history.txt"
# Initialize file with a header so the first epoch produces a complete file
with open(mape_history_path, "w", encoding="utf-8") as f:
    f.write("epoch\ttrain_mape\tval_mape\n")

for iteration_idx in range(1):

    #0.1, 0.01 no
    optimizer = torch.optim.Adam(model.parameters(), lr=0.0003) #, weight_decay=5e-4)

    # Learning rate scheduler - reduces LR by factor of gamma every step_size epochs
    scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=500, gamma=1)

    print(f"Starting training (iteration {iteration_idx + 1}, criterion '{criterion_name}')...")
    print(f"Initial learning rate: {optimizer.param_groups[0]['lr']:.6f}")
    for epoch in range(1, num_epochs + 1):
        train_loss, train_mae, train_mape = train()
        val_loss, val_mae, val_mape = test(val_loader)
        avg_mape = 0.5 * (train_mape + val_mape)

        train_losses.append(train_loss)
        val_losses.append(val_loss)
        train_maes.append(train_mae)
        val_maes.append(val_mae)
        train_mapes.append(train_mape)
        val_mapes.append(val_mape)

        # Append this epoch's train/val MAPE to the history file.
        # Open in append mode so the file is updated after every epoch.
        with open(mape_history_path, "a", encoding="utf-8") as f:
            f.write(f"{epoch}\t{train_mape:.6f}\t{val_mape:.6f}\n")

        # Save a checkpoint every time validation MAPE improves.
        # Keep all improved checkpoints (never overwrite previous bests).
        saved_best_val_mape = False
        if val_mape < best_val_mape:
            best_val_mape = val_mape
            val_mape_tag = f"{val_mape:.4f}".replace('.', 'p')
            train_mape_tag = f"{train_mape:.4f}".replace('.', 'p')
            best_mape_model_path = (
                f"best_mape_iter_{iteration_idx + 1:02d}_epoch_{epoch:04d}"
                f"_valmape_{val_mape_tag}_trainmape_{train_mape_tag}.pth"
            )
            torch.save(model.state_dict(), run_output_dir / best_mape_model_path)
            saved_best_val_mape = True

        # Save a checkpoint for the best average MAPE = (train_mape + val_mape) / 2
        saved_best_avg_mape = False
        if avg_mape < best_avg_mape:
            best_avg_mape = avg_mape
            avg_mape_tag = f"{avg_mape:.4f}".replace('.', 'p')
            val_mape_tag = f"{val_mape:.4f}".replace('.', 'p')
            train_mape_tag = f"{train_mape:.4f}".replace('.', 'p')
            best_avg_mape_model_path = (
                f"best_avgmape_iter_{iteration_idx + 1:02d}_epoch_{epoch:04d}"
                f"_avgmape_{avg_mape_tag}_valmape_{val_mape_tag}_trainmape_{train_mape_tag}.pth"
            )
            torch.save(model.state_dict(), run_output_dir / best_avg_mape_model_path)
            saved_best_avg_mape = True

        # Track best model on validation loss
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            # torch.save(model.state_dict(), 'best_model.pth')

        if epoch % 10 == 0:
            status_parts = []
            if saved_best_val_mape:
                status_parts.append('Best val MAPE checkpoint saved')
            if saved_best_avg_mape:
                status_parts.append('Best avg MAPE checkpoint saved')
            status = f" - {' | '.join(status_parts)}" if status_parts else ""
            print(f'Epoch {epoch:05d}, Train Loss: {train_loss:.6e}, Train MAE: {train_mae:.6e}, Train MAPE: {train_mape:.2f}%, Val Loss: {val_loss:.6e}, Val MAE: {val_mae:.6e}, Val MAPE: {val_mape:.2f}%, Avg MAPE: {avg_mape:.2f}%, LR: {optimizer.param_groups[0]["lr"]:.6f}{status}')

        # Step the learning rate scheduler
        scheduler.step()

    # Save the last model after training completes
    torch.save(model.state_dict(), run_output_dir / 'last_model.pth')

    print(f"Final train MAE: {train_maes[-1]:.6e}, Final train MAPE: {train_mapes[-1]:.2f}%")
    print(f"Final val MAE: {val_maes[-1]:.6e}, Final val MAPE: {val_mapes[-1]:.2f}%")
    print(f"\nTraining complete!")
    print(f"Best val loss (MSE): {best_val_loss:.6e}")
    print(f"Best val MAPE: {best_val_mape:.2f}%")
    print(f"Best avg MAPE: {best_avg_mape:.2f}%")
    print(f"Final train MAE: {train_maes[-1]:.6e}")
    print(f"Final val MAE: {val_maes[-1]:.6e}")
    print(f"Models saved in: {run_output_dir}")
    print(f"  - best_mape_iter_XX_epoch_XXXX_valmape_XXXX_trainmape_XXXX.pth")
    print(f"  - best_avgmape_iter_XX_epoch_XXXX_avgmape_XXXX_valmape_XXXX_trainmape_XXXX.pth")
    print(f"  - last_model.pth")
    print(f"  - mape_history.txt (per-epoch train/val MAPE)")




```

## 6. Evaluate convergence and predictive accuracy

Plot the training histories, evaluate the final model on both subsets, and transform predictions back to the original viscosity scale.
```python
# Generate plots for each refrigerant: pressure vs predicted viscosity, with lines for each temperature
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Get refrigerant names for training samples
train_refrigerants = df.iloc[train_indices]['Name'].values

# Create dataframe with all training data
train_df = pd.DataFrame({
    'refrigerant': train_refrigerants,
    'pressure': train_pressures,
    'temperature': train_temperatures,
    'predicted_viscosity': train_predictions,
    'actual_viscosity': train_actuals
})

# Get unique refrigerants
unique_refrigerants = sorted(train_df['refrigerant'].unique())

print(f"Creating plots for {len(unique_refrigerants)} refrigerants in the training set...")

# Calculate grid dimensions
n_refrigerants = len(unique_refrigerants)
n_cols = 4  # Number of columns in grid
n_rows = int(np.ceil(n_refrigerants / n_cols))

# Create figure with subplots in a grid
fig, axes = plt.subplots(n_rows, n_cols, figsize=(12, 3 * n_rows))
axes = axes.flatten()  # Flatten to 1D array for easier indexing

# Create a plot for each refrigerant
for idx, refrigerant in enumerate(unique_refrigerants):
    ax = axes[idx]
    
    # Filter data for this refrigerant
    ref_data = train_df[train_df['refrigerant'] == refrigerant].copy()
    
    if len(ref_data) == 0:
        ax.axis('off')
        continue
    
    # Calculate MAE for this refrigerant
    ref_mae = np.mean(np.abs(ref_data['predicted_viscosity'] - ref_data['actual_viscosity']))
    
    # Get unique temperatures and sort them
    temperatures = sorted(ref_data['temperature'].unique())
    
    # Plot a line for each temperature
    for temp in temperatures:
        temp_data = ref_data[ref_data['temperature'] == temp].sort_values('pressure')
        
        if len(temp_data) > 0:
            # Plot predicted and actual with same color
            line_pred = ax.plot(temp_data['pressure'], temp_data['predicted_viscosity'], 
                   marker='o', markersize=3, label=f'{temp:.1f} K (pred)', 
                   linewidth=1.5, linestyle='-', alpha=0.8)
            # Plot actual values (ground truth) with same color as prediction
            ax.plot(temp_data['pressure'], temp_data['actual_viscosity'], 
                   marker='s', markersize=3, label=f'{temp:.1f} K (actual)', 
                   linewidth=1.5, linestyle='--', alpha=0.6, color=line_pred[0].get_color())
    
    ax.set_xlabel('Pressure (Pa)', fontsize=9)
    ax.set_ylabel('Viscosity (PaÂ·s)', fontsize=9)
    ax.set_title(f'{refrigerant}\nMAE: {ref_mae:.1f} PaÂ·s', fontsize=10, fontweight='bold')
    ax.legend(title='Temp', fontsize=6, title_fontsize=7, loc='best', ncol=2)
    ax.grid(True, alpha=0.3)
    # ax.set_xscale('log')
    # ax.set_yscale('log')
    ax.tick_params(labelsize=8)

# Hide any unused subplots
for idx in range(n_refrigerants, len(axes)):
    axes[idx].axis('off')

plt.tight_layout()

plt.show()
print("All plots generated in grid!")
    
```
