# Evaluate a Trained Refrigerant-Viscosity GNN

This notebook reconstructs the preprocessing pipeline used in `01_train_model.ipynb`, restores a trained graph neural network, and evaluates viscosity predictions on the original refrigerant-level training and validation splits. It reports aggregate errors and produces parity and per-refrigerant isotherm plots.

> **Checkpoint selection:** By default, the notebook inspects the lexicographically latest folder under `training_runs/` and selects the last matching average-MAPE checkpoint. Confirm the printed run and checkpoint names before interpreting the results.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
import torch.nn.functional as F
from torch_geometric.loader import DataLoader
from torch_geometric.nn import GATConv, global_mean_pool
```

## 1. Resolve inputs and select a checkpoint

Locate the persisted dataset and training-run directory relative to the current working directory. Checkpoint preference is `best_avgmape*.pth`, then `best_mape*.pth`, and finally `last_model.pth`.


```python
NOTEBOOK_DIR = Path.cwd().resolve()
DATA_PATH = NOTEBOOK_DIR / "data" / "fluids_data.pkl"
TRAINING_RUNS_DIR = NOTEBOOK_DIR / "training_runs"

if not DATA_PATH.exists():
    raise FileNotFoundError(f"Dataset not found: {DATA_PATH}")

if not TRAINING_RUNS_DIR.exists():
    raise FileNotFoundError(f"Training runs directory not found: {TRAINING_RUNS_DIR}")

selected_run_dir = TRAINING_RUNS_DIR / "max_msre_20260623_001251"
checkpoint_candidates = sorted(selected_run_dir.glob("best_avgmape*.pth"))
if not checkpoint_candidates:
    checkpoint_candidates = sorted(selected_run_dir.glob("best_mape*.pth"))
if not checkpoint_candidates and (selected_run_dir / "last_model.pth").exists():
    checkpoint_candidates = [selected_run_dir / "last_model.pth"]
if not checkpoint_candidates:
    raise FileNotFoundError(f"No compatible checkpoint found in: {selected_run_dir}")

CHECKPOINT_PATH = checkpoint_candidates[-1]

print(f"Notebook directory: {NOTEBOOK_DIR}")
print(f"Dataset: {DATA_PATH.name}")
print(f"Selected run: {selected_run_dir.name}")
print(f"Checkpoint: {CHECKPOINT_PATH.name}")
```

## 2. Rebuild the evaluation data pipeline

Validate the dataset, remove graph-level descriptors that are zero for every sample, append pressure and temperature, and recreate the exact refrigerant holdout used during training. All scaling parameters are fitted on the training subset only, matching the training workflow and preventing validation leakage.


```python
df = pd.read_pickle(DATA_PATH)

df = df[df["Name"].str.startswith("R", na=False)].reset_index(drop=True)

required_cols = ["graph", "pressure", "temperature", "viscosity"]
missing_cols = [column for column in required_cols if column not in df.columns]
if missing_cols:
    raise KeyError(f"Missing required columns in dataframe: {missing_cols}")

for column in ["pressure", "temperature", "viscosity"]:
    df[column] = pd.to_numeric(df[column], errors="coerce")

df = df.dropna(subset=required_cols).reset_index(drop=True)

negative_visc = df[df["viscosity"] < 0]
if not negative_visc.empty:
    raise ValueError("Found negative viscosity values in the dataset.")

print(f"Samples retained: {len(df)}")
print(f"Refrigerants retained: {len(df['Name'].unique())}")
print(f"Viscosity range: {df['viscosity'].min():.6e} to {df['viscosity'].max():.6e}")
```

```python
n_attributes = len(df.iloc[0]["graph"].u[0])
all_zero_attributes = []

for attr_idx in range(n_attributes):
    is_all_zero = True
    for _, row in df.iterrows():
        graph = row["graph"]
        attr_values = [graph.u[node_idx][attr_idx] for node_idx in range(len(graph.u))]
        if any(value != 0 for value in attr_values):
            is_all_zero = False
            break
    if is_all_zero:
        all_zero_attributes.append(attr_idx)

all_zero_set = set(all_zero_attributes)
indices_to_keep = [idx for idx in range(n_attributes) if idx not in all_zero_set]

data_list = []
for _, row in df.iterrows():
    graph = row["graph"].clone()
    if all_zero_attributes and len(indices_to_keep) < n_attributes:
        current_attr_count = graph.u.shape[1] if len(graph.u.shape) == 2 else len(graph.u[0])
        if current_attr_count == n_attributes:
            graph.u = graph.u[:, indices_to_keep]

    pressure_temp = torch.tensor(
        [float(row["pressure"]), float(row["temperature"])],
        dtype=torch.float32,
    ).view(1, -1)

    if hasattr(graph, "u") and graph.u is not None:
        graph.u = torch.cat([graph.u, pressure_temp], dim=1)
    else:
        graph.u = pressure_temp

    graph.y = torch.tensor([[float(row["viscosity"])]], dtype=torch.float32)
    data_list.append(graph)

print(f"Dropped all-zero global attributes: {all_zero_attributes}")
print(f"Total graph samples: {len(data_list)}")
print(f"Node feature shape: {data_list[0].x.shape}")
print(f"Global feature shape: {data_list[0].u.shape}")
print(f"Target shape: {data_list[0].y.shape}")
```

```python
val_refrigerants = ["R12", "R124", "R125", "R1234zeE", "R218"]
train_refrigerants = [name for name in df["Name"].unique() if name not in val_refrigerants]

train_indices = df[df["Name"].isin(train_refrigerants)].index.to_numpy()
val_indices = df[df["Name"].isin(val_refrigerants)].index.to_numpy()

train_data = [data_list[idx] for idx in train_indices]
val_data = [data_list[idx] for idx in val_indices]

if set(train_refrigerants).intersection(val_refrigerants):
    raise RuntimeError("Train/validation refrigerant overlap detected.")
if set(train_indices).intersection(val_indices):
    raise RuntimeError("Train/validation sample overlap detected.")

train_node_features = torch.cat([data.x for data in train_data], dim=0)
node_min = train_node_features.min(dim=0, keepdim=True).values
node_max = train_node_features.max(dim=0, keepdim=True).values
node_range = torch.where(node_max - node_min > 1e-8, node_max - node_min, torch.ones_like(node_max - node_min))

for data in train_data:
    data.x = (data.x - node_min) / node_range
for data in val_data:
    data.x = (data.x - node_min) / node_range

if hasattr(train_data[0], "edge_attr") and train_data[0].edge_attr is not None:
    train_edge_features = torch.cat([data.edge_attr for data in train_data if data.edge_attr is not None], dim=0)
    edge_min = train_edge_features.min(dim=0, keepdim=True).values
    edge_max = train_edge_features.max(dim=0, keepdim=True).values
    edge_range = torch.where(edge_max - edge_min > 1e-8, edge_max - edge_min, torch.ones_like(edge_max - edge_min))

    for data in train_data:
        if data.edge_attr is not None:
            data.edge_attr = (data.edge_attr - edge_min) / edge_range
    for data in val_data:
        if data.edge_attr is not None:
            data.edge_attr = (data.edge_attr - edge_min) / edge_range
else:
    edge_min = None
    edge_max = None
    edge_range = None

train_global_features = torch.cat([data.u for data in train_data], dim=0)
global_min = train_global_features.min(dim=0, keepdim=True).values
global_max = train_global_features.max(dim=0, keepdim=True).values
global_range = torch.where(global_max - global_min > 1e-8, global_max - global_min, torch.ones_like(global_max - global_min))

for data in train_data:
    data.u = (data.u - global_min) / global_range
for data in val_data:
    data.u = (data.u - global_min) / global_range

y_min = torch.tensor(0.0)
train_output_values = torch.cat([data.y for data in train_data], dim=0)
y_max = train_output_values.max()
y_range = y_max - y_min
if y_range < 1e-8:
    y_range = torch.tensor(1.0)

for data in train_data:
    data.y = (data.y - y_min) / y_range
for data in val_data:
    data.y = (data.y - y_min) / y_range

batch_size = 32
train_loader = DataLoader(train_data, batch_size=batch_size, shuffle=False)
val_loader = DataLoader(val_data, batch_size=batch_size, shuffle=False)

print(f"Training refrigerants: {sorted(train_refrigerants)}")
print(f"Validation refrigerants: {sorted(val_refrigerants)}")
print(f"Training samples: {len(train_data)}")
print(f"Validation samples: {len(val_data)}")
```

```python
class GNN(torch.nn.Module):
    def __init__(self, node_features, hidden_dim, global_features=2, num_outputs=1):
        super().__init__()
        self.conv1 = GATConv(node_features, hidden_dim, heads=4, concat=True)
        self.conv2 = GATConv(hidden_dim * 4, hidden_dim, heads=4, concat=True)
        self.conv3 = GATConv(hidden_dim * 4, hidden_dim, heads=4, concat=True)
        self.global_fc = torch.nn.Linear(global_features, hidden_dim)
        self.fc1 = torch.nn.Linear(hidden_dim * 4 + hidden_dim, hidden_dim)
        self.fc2 = torch.nn.Linear(hidden_dim, num_outputs)
        self.dropout = torch.nn.Dropout(0.2)

    def forward(self, x, edge_index, batch, u):
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = self.dropout(x)

        x = self.conv2(x, edge_index)
        x = F.relu(x)
        x = self.dropout(x)

        x = self.conv3(x, edge_index)
        x = F.relu(x)
        x = global_mean_pool(x, batch)

        u = F.relu(self.global_fc(u))
        x = torch.cat([x, u], dim=1)
        x = F.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        x = F.softplus(x)
        return x

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
node_features = data_list[0].x.shape[1]
global_features = data_list[0].u.shape[1]
num_outputs = data_list[0].y.shape[1]
hidden_dim = 128

model = GNN(node_features, hidden_dim, global_features, num_outputs).to(device)
state_dict = torch.load(CHECKPOINT_PATH, map_location=device)
model.load_state_dict(state_dict)
model.eval()

print(f"Model restored on {device}")
print(f"Checkpoint loaded: {CHECKPOINT_PATH.name}")
print(f"Total parameters: {sum(parameter.numel() for parameter in model.parameters())}")
```

## 3. Restore the model and run inference

Instantiate the same graph-attention architecture used for training, load the selected state dictionary, and evaluate batches without gradient tracking. Predictions, targets, pressure, and temperature are transformed back to their original scales before MAE and MAPE are calculated.


```python
def collect_predictions(loader):
    predictions = []
    actuals = []
    pressures = []
    temperatures = []

    with torch.no_grad():
        for batch in loader:
            batch = batch.to(device)
            output = model(batch.x, batch.edge_index, batch.batch, batch.u)
            predictions.append(output.cpu().numpy())
            actuals.append(batch.y.cpu().numpy())
            pressures.append(batch.u[:, -2].cpu().numpy())
            temperatures.append(batch.u[:, -1].cpu().numpy())

    predictions = np.vstack(predictions).reshape(-1)
    actuals = np.vstack(actuals).reshape(-1)
    pressures = np.concatenate(pressures)
    temperatures = np.concatenate(temperatures)

    predictions = predictions * y_range.item() + y_min.item()
    actuals = actuals * y_range.item() + y_min.item()
    pressures = pressures * global_range[0, -2].item() + global_min[0, -2].item()
    temperatures = temperatures * global_range[0, -1].item() + global_min[0, -1].item()

    return {
        "pred_visc": predictions,
        "actual_visc": actuals,
        "pressure": pressures,
        "temperature": temperatures,
    }

train_results = collect_predictions(train_loader)
val_results = collect_predictions(val_loader)

for split_name, results in [("Train", train_results), ("Validation", val_results)]:
    mae_visc = np.mean(np.abs(results["pred_visc"] - results["actual_visc"]))
    mape_visc = np.mean(
        np.abs(
            (results["pred_visc"] - results["actual_visc"])
            / np.maximum(np.abs(results["actual_visc"]), 1e-12)
        )
    ) * 100
    print(f"{split_name} viscosity: MAE={mae_visc:.6e}, MAPE={mape_visc:.2f}%")
```

## 4. Visual diagnostics and summary

The logâ€“log parity plot compares all predictions with the ideal identity line. The subsequent panels show predicted and reference pressure sweeps at each sampled temperature, with per-fluid MAPE in each title. Training panels diagnose fit quality; validation panels assess transfer to molecular graphs not seen during optimization.


```python
# Parity plot: predicted vs actual viscosity
fig, ax = plt.subplots(figsize=(7, 7))

ax.scatter(
    train_results["actual_visc"],
    train_results["pred_visc"],
    s=18,
    alpha=0.6,
    label="Train",
)
ax.scatter(
    val_results["actual_visc"],
    val_results["pred_visc"],
    s=24,
    alpha=0.8,
    label="Validation",
)

all_visc = np.concatenate(
    [
        train_results["actual_visc"],
        val_results["actual_visc"],
        train_results["pred_visc"],
        val_results["pred_visc"],
    ]
)
vmin, vmax = np.min(all_visc), np.max(all_visc)
ax.plot([vmin, vmax], [vmin, vmax], "k--", linewidth=1.5, label="Ideal y=x")
ax.set_xlim(vmin, vmax)
ax.set_ylim(vmin, vmax)
ax.set_xlabel("Actual viscosity")
ax.set_ylabel("Predicted viscosity")
ax.set_title("Parity Plot - Viscosity")
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_xscale("log")
ax.set_yscale("log")

plt.tight_layout()
plt.show()
```

```python
# Generate plots for each refrigerant in the TRAINING set
train_refrigerants_names = df.iloc[train_indices]["Name"].values

train_df = pd.DataFrame(
    {
        "refrigerant": train_refrigerants_names,
        "pressure": train_results["pressure"],
        "temperature": train_results["temperature"],
        "predicted_viscosity": train_results["pred_visc"],
        "actual_viscosity": train_results["actual_visc"],
    }
)

unique_refrigerants = sorted(train_df["refrigerant"].unique())
print(f"Creating plots for {len(unique_refrigerants)} refrigerants in the training set...")

n_refrigerants = len(unique_refrigerants)
n_cols = 4
n_rows = int(np.ceil(n_refrigerants / n_cols))

fig, axes = plt.subplots(n_rows, n_cols, figsize=(12, 3 * n_rows))
if n_refrigerants == 1:
    axes = np.array([axes])
axes = axes.flatten()
fig.suptitle("Training Set - Viscosity", fontsize=14, fontweight="bold", y=1.00)

legend_handles = []
legend_labels = []

for idx, refrigerant in enumerate(unique_refrigerants):
    ax = axes[idx]
    ref_data = train_df[train_df["refrigerant"] == refrigerant]
    if len(ref_data) == 0:
        ax.axis("off")
        continue

    ref_mape = np.mean(
        np.abs(
            (ref_data["predicted_viscosity"] - ref_data["actual_viscosity"])
            / np.maximum(np.abs(ref_data["actual_viscosity"]), 1e-12)
        )
    ) * 100
    temperatures = sorted(ref_data["temperature"].unique())

    for temp in temperatures:
        temp_data = ref_data[ref_data["temperature"] == temp].sort_values("pressure")
        if len(temp_data) > 0:
            line_pred = ax.plot(
                temp_data["pressure"],
                temp_data["predicted_viscosity"],
                marker="o",
                markersize=3,
                label=f"{temp:.1f} K (pred)",
                linewidth=1.5,
                linestyle="-",
                alpha=0.8,
            )
            ax.plot(
                temp_data["pressure"],
                temp_data["actual_viscosity"],
                marker="s",
                markersize=3,
                label=f"{temp:.1f} K (actual)",
                linewidth=1.5,
                linestyle="--",
                alpha=0.6,
                color=line_pred[0].get_color(),
            )

    if not legend_labels:
        legend_handles, legend_labels = ax.get_legend_handles_labels()

    ax.set_xlabel("Pressure (Pa)", fontsize=9)
    ax.set_ylabel("Viscosity", fontsize=9)
    ax.set_title(
        f"{refrigerant}\nMAPE: {ref_mape:.2f}%", fontsize=10, fontweight="bold"
    )
    ax.grid(True, alpha=0.3)
    ax.tick_params(labelsize=8)

for idx in range(n_refrigerants, len(axes)):
    axes[idx].axis("off")

if legend_labels:
    unique_legend = {}
    for handle, label in zip(legend_handles, legend_labels):
        if label not in unique_legend:
            unique_legend[label] = handle
    fig.legend(
        list(unique_legend.values()),
        list(unique_legend.keys()),
        title="Temp",
        fontsize=6,
        title_fontsize=7,
        loc="upper center",
        ncol=min(6, len(unique_legend)),
        bbox_to_anchor=(0.5, 0.99),
    )

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()

print("All training viscosity plots generated.")
```

```python
# Generate plots for each refrigerant in the VALIDATION set
val_refrigerants_names = df.iloc[val_indices]["Name"].values

val_df = pd.DataFrame(
    {
        "refrigerant": val_refrigerants_names,
        "pressure": val_results["pressure"],
        "temperature": val_results["temperature"],
        "predicted_viscosity": val_results["pred_visc"],
        "actual_viscosity": val_results["actual_visc"],
    }
)

unique_val_refrigerants = sorted(val_df["refrigerant"].unique())
print(
    f"Creating plots for {len(unique_val_refrigerants)} refrigerants in the validation set..."
)

n_val_refrigerants = len(unique_val_refrigerants)
n_cols = 4
n_rows = int(np.ceil(n_val_refrigerants / n_cols))

fig, axes = plt.subplots(n_rows, n_cols, figsize=(12, 3 * n_rows))
if n_val_refrigerants == 1:
    axes = np.array([axes])
axes = axes.flatten()
fig.suptitle("Validation Set - Viscosity", fontsize=14, fontweight="bold", y=1.00)

legend_handles = []
legend_labels = []

for idx, refrigerant in enumerate(unique_val_refrigerants):
    ax = axes[idx]
    ref_data = val_df[val_df["refrigerant"] == refrigerant]
    if len(ref_data) == 0:
        ax.axis("off")
        continue

    ref_mape = np.mean(
        np.abs(
            (ref_data["predicted_viscosity"] - ref_data["actual_viscosity"])
            / np.maximum(np.abs(ref_data["actual_viscosity"]), 1e-12)
        )
    ) * 100
    temperatures = sorted(ref_data["temperature"].unique())

    for temp in temperatures:
        temp_data = ref_data[ref_data["temperature"] == temp].sort_values("pressure")
        if len(temp_data) > 0:
            line_pred = ax.plot(
                temp_data["pressure"],
                temp_data["predicted_viscosity"],
                marker="o",
                markersize=3,
                label=f"{temp:.1f} K (pred)",
                linewidth=1.5,
                linestyle="-",
                alpha=0.8,
            )
            ax.plot(
                temp_data["pressure"],
                temp_data["actual_viscosity"],
                marker="s",
                markersize=3,
                label=f"{temp:.1f} K (actual)",
                linewidth=1.5,
                linestyle="--",
                alpha=0.6,
                color=line_pred[0].get_color(),
            )

    if not legend_labels:
        legend_handles, legend_labels = ax.get_legend_handles_labels()

    ax.set_xlabel("Pressure (Pa)", fontsize=9)
    ax.set_ylabel("Viscosity", fontsize=9)
    ax.set_title(
        f"{refrigerant} [VAL]\nMAPE: {ref_mape:.2f}%", fontsize=10, fontweight="bold"
    )
    ax.grid(True, alpha=0.3)
    ax.tick_params(labelsize=8)

for idx in range(n_val_refrigerants, len(axes)):
    axes[idx].axis("off")

if legend_labels:
    unique_legend = {}
    for handle, label in zip(legend_handles, legend_labels):
        if label not in unique_legend:
            unique_legend[label] = handle
    fig.legend(
        list(unique_legend.values()),
        list(unique_legend.keys()),
        title="Temp",
        fontsize=6,
        title_fontsize=7,
        loc="upper center",
        ncol=min(6, len(unique_legend)),
        bbox_to_anchor=(0.5, 0.99),
    )

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()

print("All validation viscosity plots generated.")
```
