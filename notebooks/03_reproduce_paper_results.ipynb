# Publication Results and Figures

This notebook reproduces the quantitative tables and publication-ready figures derived from four completed training runs. It reconstructs the evaluation pipeline, restores the best MSE, MSRE, MAPE, and Max-MSRE checkpoints, compares their predictive performance, and treats Max-MSRE as the primary model for detailed diagnostics.

Main outputs include the dataset and refrigerant-classification tables, loss-function comparison, per-refrigerant errors, convergence and parity figures, permutation feature importance, and appendix plots. Tabular and graphical artifacts are written beneath `results/`.

> **Reproducibility note:** Run folders and the primary checkpoint are pinned explicitly below. Update those paths deliberately when regenerating results from newer experiments.

```python
# Load and prepare data (same pipeline as 1_Training)
df = pd.read_pickle(DATA_PATH)
df = df[df['Name'].str.startswith('R', na=False)].reset_index(drop=True)

for col in ['pressure', 'temperature', 'viscosity']:
    df[col] = pd.to_numeric(df[col], errors='coerce')
df = df.dropna(subset=['graph', 'pressure', 'temperature', 'viscosity']).reset_index(drop=True)

# Drop all-zero global attributes
n_attributes = len(df.iloc[0]['graph'].u[0])
all_zero_attributes = []
for attr_idx in range(n_attributes):
    is_all_zero = True
    for _, row in df.iterrows():
        graph = row['graph']
        attr_values = [graph.u[node_idx][attr_idx] for node_idx in range(len(graph.u))]
        if any(val != 0 for val in attr_values):
            is_all_zero = False
            break
    if is_all_zero:
        all_zero_attributes.append(attr_idx)

all_zero_set = set(all_zero_attributes)
indices_to_keep = [i for i in range(n_attributes) if i not in all_zero_set]

# Build data_list with global features + pressure/temperature
data_list = []
for _, row in df.iterrows():
    graph = row['graph'].clone()
    if all_zero_attributes and len(indices_to_keep) < n_attributes:
        current_count = graph.u.shape[1] if len(graph.u.shape) == 2 else len(graph.u[0])
        if current_count == n_attributes:
            graph.u = graph.u[:, indices_to_keep]
    pressure_temp = torch.tensor([float(row['pressure']), float(row['temperature'])], dtype=torch.float).view(1, -1)
    if hasattr(graph, 'u') and graph.u is not None:
        graph.u = torch.cat([graph.u, pressure_temp], dim=1)
    else:
        graph.u = pressure_temp
    graph.y = torch.tensor([float(row['viscosity'])], dtype=torch.float).view(1, 1)
    data_list.append(graph)

print(f"Dataset: {len(df)} samples, {df['Name'].nunique()} refrigerants")
print(f"Node features: {data_list[0].x.shape[1]}")
print(f"Global features: {data_list[0].u.shape[1]}")
print(f"Dropped {len(all_zero_attributes)} all-zero global attributes")
```

```python
# Train/validation split (same as training)
val_refrigerants = ['R12', 'R124', 'R125', 'R1234zeE', 'R218']
train_refrigerants = [r for r in df['Name'].unique() if r not in val_refrigerants]

train_indices = df[df['Name'].isin(train_refrigerants)].index.to_numpy()
val_indices = df[df['Name'].isin(val_refrigerants)].index.to_numpy()

train_data = [data_list[i] for i in train_indices]
val_data = [data_list[i] for i in val_indices]

# Normalize using training statistics
train_node_features = torch.cat([d.x for d in train_data], dim=0)
node_min = train_node_features.min(dim=0, keepdim=True).values
node_max = train_node_features.max(dim=0, keepdim=True).values
node_range = torch.where(node_max - node_min > 1e-8, node_max - node_min, torch.ones_like(node_max))

if hasattr(train_data[0], 'edge_attr') and train_data[0].edge_attr is not None:
    train_edge_features = torch.cat([d.edge_attr for d in train_data if d.edge_attr is not None], dim=0)
    edge_min = train_edge_features.min(dim=0, keepdim=True).values
    edge_max = train_edge_features.max(dim=0, keepdim=True).values
    edge_range = torch.where(edge_max - edge_min > 1e-8, edge_max - edge_min, torch.ones_like(edge_max))
else:
    edge_min = edge_max = edge_range = None

train_global_features = torch.cat([d.u for d in train_data], dim=0)
global_min = train_global_features.min(dim=0, keepdim=True).values
global_max = train_global_features.max(dim=0, keepdim=True).values
global_range = torch.where(global_max - global_min > 1e-8, global_max - global_min, torch.ones_like(global_max))

y_min = torch.tensor(0.0)
train_output_values = torch.cat([d.y for d in train_data], dim=0)
y_max = train_output_values.max()
y_range = y_max - y_min
if y_range < 1e-8:
    y_range = torch.tensor(1.0)

# Apply normalization
for data in train_data + val_data:
    data.x = (data.x - node_min) / node_range
    if edge_range is not None and data.edge_attr is not None:
        data.edge_attr = (data.edge_attr - edge_min) / edge_range
    data.u = (data.u - global_min) / global_range
    data.y = (data.y - y_min) / y_range

batch_size = 32
train_loader = DataLoader(train_data, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_data, batch_size=batch_size, shuffle=False)

print(f"Train: {len(train_data)} samples ({len(train_refrigerants)} refrigerants)")
print(f"Val: {len(val_data)} samples ({len(val_refrigerants)} refrigerants)")
print(f"y_range = {y_range.item():.2f} ÂµPaÂ·s")
```

Report the thermodynamic coverage, sample counts, chemical-class coverage, and graph feature dimensions used by every model comparison in this notebook.

```python
# Table 1: Dataset summary
T_range = (df['temperature'].min(), df['temperature'].max())
P_range = (df['pressure'].min() / 1e6, df['pressure'].max() / 1e6)  # MPa
visc_range = (df['viscosity'].min(), df['viscosity'].max())

class_counts = df['Name'].map(lambda x: next((v for k, v in {
    'CFC': ['R11','R113','R115','R12','R13'],
    'HCFC': ['R123','R124','R141b','R142b','R21','R22'],
    'HFC': ['R125','R134a','R143a','R152a','R227ea','R23','R236ea','R236fa','R245fa','R32','R41'],
    'HFO': ['R1234yf','R1234zeE'],
    'PFC': ['R116','R14','R218','RC318'],
    'HCFO': ['R1233zdE'],
}.items() if x in v), None))

print("\n" + "="*70)
print("TABLE 1: Dataset Summary")
print("="*70)
print(f"Total refrigerants:       {df['Name'].nunique()}")
print(f"Total data points:        {len(df)}")
print(f"Chemical classes:         6 (CFC, HCFC, HFC, HFO, PFC, HCFO)")
print(f"Temperature range:        {T_range[0]:.0f} -- {T_range[1]:.0f} K")
print(f"Pressure range:           {P_range[0]:.1f} -- {P_range[1]:.1f} MPa")
print(f"Viscosity range:          {visc_range[0]:.1f} -- {visc_range[1]:.1f} ÂµPaÂ·s")
print(f"Points per refrigerant:   {len(df) // df['Name'].nunique()}")
print(f"Node features per atom:   {data_list[0].x.shape[1]}")
print(f"Edge features per bond:   {data_list[0].edge_attr.shape[1] if data_list[0].edge_attr is not None else 0}")
print(f"Global features:          {data_list[0].u.shape[1]}")
print("="*70)
```

```python
# Model definition (same as 1_Training.ipynb)
class GNN(torch.nn.Module):
    def __init__(self, node_features, hidden_dim, global_features=2):
        super(GNN, self).__init__()

        # Graph attention layers (multi-head with concat=True: output size = hidden_dim * heads)
        self.conv1 = GATConv(node_features, hidden_dim, heads=4, concat=True)
        self.conv2 = GATConv(hidden_dim * 4, hidden_dim, heads=4, concat=True)
        self.conv3 = GATConv(hidden_dim * 4, hidden_dim, heads=4, concat=True)

        # Process global features (pressure/temperature)
        self.global_fc = torch.nn.Linear(global_features, hidden_dim)

        # Final prediction layers (input: hidden_dim*4 from conv3 + hidden_dim from global_fc = hidden_dim*5)
        self.fc1 = torch.nn.Linear(hidden_dim * 4 + hidden_dim, hidden_dim)
        self.fc2 = torch.nn.Linear(hidden_dim, 1)

        self.dropout = torch.nn.Dropout(0.2)

    def forward(self, x, edge_index, batch, u):
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = self.dropout(x)

        x = self.conv2(x, edge_index)
        x = F.relu(x)
        x = self.dropout(x)

        x = self.conv3(x, edge_index)
        x = F.relu(x)

        # Graph-level embedding
        x = global_mean_pool(x, batch)

        # Global features (pressure/temperature) embedding
        u = F.relu(self.global_fc(u))

        # Fusion + scalar viscosity output
        x = torch.cat([x, u], dim=1)
        x = F.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)

        # Ensure positive output (viscosity is always positive)
        x = F.softplus(x)

        return x


device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
node_features = data_list[0].x.shape[1]
global_features = data_list[0].u.shape[1]
hidden_dim = 128

print(f"Device: {device}")
print(f"Architecture: GATConvÃ—3 (4 heads, hidden={hidden_dim}) + global FC")
print(f"Node features: {node_features}, Global features: {global_features}")
```

## 1. Load training histories and best checkpoints (4 loss criteria)

The four loss criteria (MSE, MSRE, MAPE, Max-MSRE) were each trained in
`1_Training.ipynb`. For every run we load:

* `mape_history.txt` -- per-epoch train/val MAPE (used for the evolution figure)
* the highest-epoch `best_avgmape_*.pth` checkpoint (used for the metric table)

The Max-MSRE checkpoint is the *primary* model: it is used for the
per-refrigerant table, the parity plot and the permutation feature importance.

```python
# ----------------------------------------------------------------------------
# Load per-epoch MAPE histories for the 4 criteria.
# Each file has 3 tab-separated columns: epoch / train_mape / val_mape.
# ----------------------------------------------------------------------------
mape_histories = {}
for name, run_dir in CRITERION_RUNS.items():
    history_path = run_dir / "mape_history.txt"
    mape_histories[name] = pd.read_csv(history_path, sep="\t")
    last = mape_histories[name].iloc[-1]
    print(f"{name:<10s} {len(mape_histories[name]):5d} epochs | "
          f"final train MAPE = {last['train_mape']:7.3f}% | "
          f"final val MAPE = {last['val_mape']:7.3f}%")
```

```python
import re

# ----------------------------------------------------------------------------
# Pick the best checkpoint per criterion.
# The training loop saves `best_avgmape_iter_XX_epoch_EEEE_avgmape_VpVVVV_...pth`
# *each time* the avg MAPE improves, so the file with the largest epoch index
# in a given folder is by construction the best of that run.
# For the primary criterion we use the explicit pinned checkpoint instead.
# ----------------------------------------------------------------------------
_epoch_re = re.compile(r"_epoch_(\d+)_")


def _latest_best_checkpoint(run_dir: Path) -> Path:
    candidates = list(run_dir.glob("best_avgmape_iter_*.pth"))
    if not candidates:
        raise FileNotFoundError(f"No best_avgmape checkpoint found in {run_dir}")
    def _epoch_of(p: Path) -> int:
        m = _epoch_re.search(p.name)
        return int(m.group(1)) if m else -1
    return max(candidates, key=_epoch_of)


criterion_checkpoints = {}
for name, run_dir in CRITERION_RUNS.items():
    if name == PRIMARY_CRITERION:
        criterion_checkpoints[name] = BEST_CHECKPOINT
    else:
        criterion_checkpoints[name] = _latest_best_checkpoint(run_dir)

print("Best checkpoint per criterion:")
for name, ckpt in criterion_checkpoints.items():
    print(f"  {name:<10s} {ckpt.name}")
```

## 2. Inference with the 4 best checkpoints

Load every best checkpoint and run inference on both splits. Results are
collected per criterion. The Max-MSRE model is also exposed as `model_primary`
for the downstream parity plot, per-refrigerant table and feature importance.

```python
# Load all 4 best checkpoints into individual GNN instances.
criterion_models = {}
for name, ckpt in criterion_checkpoints.items():
    m = GNN(node_features, hidden_dim, global_features).to(device)
    m.load_state_dict(torch.load(ckpt, map_location=device))
    m.eval()
    criterion_models[name] = m
    print(f"Loaded {name:<10s} from {ckpt.name}")

# Convenient alias used by the rest of the notebook.
model_primary = criterion_models[PRIMARY_CRITERION]
```

```python
@torch.no_grad()
def get_predictions(model, loader, device):
    """Run inference and return denormalized predictions and actuals."""
    model.eval()
    preds, actuals, pressures, temperatures = [], [], [], []
    for data in loader:
        data = data.to(device)
        out = model(data.x, data.edge_index, data.batch, data.u)
        preds.append(out.cpu().numpy())
        actuals.append(data.y.cpu().numpy())
        pressures.append(data.u[:, -2].cpu().numpy())
        temperatures.append(data.u[:, -1].cpu().numpy())

    preds = np.vstack(preds).reshape(-1)
    actuals = np.vstack(actuals).reshape(-1)
    pressures = np.concatenate(pressures)
    temperatures = np.concatenate(temperatures)

    # Denormalize
    preds = preds * y_range.item() + y_min.item()
    actuals = actuals * y_range.item() + y_min.item()
    pressures = pressures * global_range[0, -2].item() + global_min[0, -2].item()
    temperatures = temperatures * global_range[0, -1].item() + global_min[0, -1].item()

    return preds, actuals, pressures, temperatures


# Non-shuffled loaders for ordered inference
train_loader_ordered = DataLoader(train_data, batch_size=batch_size, shuffle=False)
val_loader_ordered = DataLoader(val_data, batch_size=batch_size, shuffle=False)

# Run inference for every criterion. Train/val P and T are identical for
# all models (they are properties of the dataset, not the model) so we
# capture them only from the primary model.
predictions = {}  # name -> {"train_pred", "train_actual", "val_pred", "val_actual"}
train_P = train_T = val_P = val_T = None

for name, model in criterion_models.items():
    tr_pred, tr_actual, tr_P, tr_T = get_predictions(model, train_loader_ordered, device)
    va_pred, va_actual, va_P, va_T = get_predictions(model, val_loader_ordered, device)
    predictions[name] = {
        "train_pred":   tr_pred,
        "train_actual": tr_actual,
        "val_pred":     va_pred,
        "val_actual":   va_actual,
    }
    if name == PRIMARY_CRITERION:
        train_P, train_T = tr_P, tr_T
        val_P, val_T = va_P, va_T

# Aliases for the primary model -- used by parity plot, per-ref table, FI.
primary_train_pred   = predictions[PRIMARY_CRITERION]["train_pred"]
primary_train_actual = predictions[PRIMARY_CRITERION]["train_actual"]
primary_val_pred     = predictions[PRIMARY_CRITERION]["val_pred"]
primary_val_actual   = predictions[PRIMARY_CRITERION]["val_actual"]

print(f"Inference complete for {len(predictions)} models.")
print(f"Train samples: {len(primary_train_pred)}, Val samples: {len(primary_val_pred)}")
```

## 3. Per-refrigerant error table (Table 4) â€” primary model only

Aggregate absolute percentage errors by refrigerant for the primary Max-MSRE model. `MAPE` summarizes typical relative error across the 25 state points, whereas `MaxAPE` records the largest pointwise error for each fluid.

```python
# Build per-refrigerant MAPE for the primary (Max-MSRE) model.
train_names = df.iloc[train_indices]['Name'].values
val_names = df.iloc[val_indices]['Name'].values

# Refrigerant class mapping
class_map = {}
for r in ['R11', 'R113', 'R115', 'R12', 'R13']:
    class_map[r] = 'CFC'
for r in ['R123', 'R124', 'R141b', 'R142b', 'R21', 'R22']:
    class_map[r] = 'HCFC'
for r in ['R125', 'R134a', 'R143a', 'R152a', 'R227ea', 'R23', 'R236ea', 'R236fa', 'R245fa', 'R32', 'R41']:
    class_map[r] = 'HFC'
for r in ['R1234yf', 'R1234zeE']:
    class_map[r] = 'HFO'
for r in ['R116', 'R14', 'R218', 'RC318']:
    class_map[r] = 'PFC'
for r in ['R1233zdE']:
    class_map[r] = 'HCFO'

# Combine training + validation predictions for the primary model.
all_names = np.concatenate([train_names, val_names])
all_actuals = np.concatenate([primary_train_actual, primary_val_actual])
all_preds = np.concatenate([primary_train_pred, primary_val_pred])
all_split = np.concatenate([
    np.full(len(train_names), 'Train'),
    np.full(len(val_names), 'Val'),
])

results_df = pd.DataFrame({
    'Name': all_names,
    'Split': all_split,
    'Actual': all_actuals,
    'Predicted': all_preds,
})
results_df['Class'] = results_df['Name'].map(class_map)
results_df['APE'] = (
    np.abs(results_df['Predicted'] - results_df['Actual'])
    / np.maximum(np.abs(results_df['Actual']), 1e-12)
    * 100
)

# Per-refrigerant MAPE
per_ref_mape = (
    results_df
    .groupby(['Class', 'Name', 'Split'])
    .agg(
        MAPE=('APE', 'mean'),
        MaxAPE=('APE', 'max'),
        N_samples=('Actual', 'count'),
    )
    .reset_index()
    .sort_values(['Class', 'Name'])
)

print("\n" + "=" * 80)
print(f"TABLE 4: Per-Refrigerant MAPE (%) -- {PRIMARY_CRITERION} model")
print("=" * 80)
print(per_ref_mape.to_string(index=False, float_format='%.2f'))
print()

per_ref_mape.to_csv(NOTEBOOK_DIR / "results" / "per_refrigerant_mape.csv", index=False)
print("Saved to results/per_refrigerant_mape.csv")
```

## 4. Loss-function comparison (Table 3) â€” four criteria

Compare the best checkpoint from each objective using train and validation MAPE, absolute-error metrics, the worst pointwise validation error, and the worst refrigerant-level MAPE. Metrics are computed after converting predictions back to $\mu$Pa$\cdot$s$.

```python
def compute_metrics(preds, actuals):
    mae = np.mean(np.abs(preds - actuals))
    rmse = np.sqrt(np.mean((preds - actuals) ** 2))
    rel = np.abs((preds - actuals) / np.maximum(np.abs(actuals), 1e-12))
    return {
        'MAE': mae,
        'RMSE': rmse,
        'MAPE': rel.mean() * 100,
        'Max APE': rel.max() * 100,
    }


def per_ref_worst_mape(names, preds, actuals):
    """Worst per-refrigerant mean APE for a given prediction array."""
    rel = (
        np.abs(preds - actuals)
        / np.maximum(np.abs(actuals), 1e-12)
        * 100
    )
    tmp = pd.DataFrame({'Name': names, 'APE': rel})
    return tmp.groupby('Name')['APE'].mean().max()


all_names_concat = np.concatenate([train_names, val_names])

# Build one column per criterion.
columns = {}
for name in CRITERION_RUNS.keys():
    p = predictions[name]
    tr_m = compute_metrics(p['train_pred'], p['train_actual'])
    va_m = compute_metrics(p['val_pred'], p['val_actual'])
    worst = per_ref_worst_mape(
        all_names_concat,
        np.concatenate([p['train_pred'], p['val_pred']]),
        np.concatenate([p['train_actual'], p['val_actual']]),
    )
    columns[name] = [
        tr_m['MAPE'],
        va_m['MAPE'],
        0.5 * (tr_m['MAPE'] + va_m['MAPE']),
        tr_m['MAE'],
        va_m['MAE'],
        va_m['RMSE'],
        va_m['Max APE'],
        worst,
    ]

comparison_table = pd.DataFrame(
    {
        'Metric': [
            'Train MAPE (%)', 'Val MAPE (%)', 'Avg MAPE (%)',
            'Train MAE (uPa.s)', 'Val MAE (uPa.s)',
            'Val RMSE (uPa.s)', 'Val Max APE (%)',
            'Worst per-ref MAPE (%)',
        ],
        **columns,
    }
)

print("\n" + "=" * 80)
print("TABLE 3: Loss Function Comparison (4 criteria)")
print("=" * 80)
print(comparison_table.to_string(index=False, float_format='%.2f'))

comparison_table.to_csv(NOTEBOOK_DIR / "results" / "loss_comparison.csv", index=False)
print("\nSaved to results/loss_comparison.csv")
```

## 8. Refrigerant classification table (Table 2)

Generate the complete list of 29 refrigerants with chemical class, molecular formula, exact molecular weight, and train/validation assignment. Formulae and molecular weights are derived from the stored SMILES with RDKit.

```python
# Build classification table from the dataset
ref_info = df[['Name', 'smiles', 'class']].drop_duplicates().reset_index(drop=True)
ref_info['Split'] = ref_info['Name'].apply(lambda x: 'Validation' if x in val_refrigerants else 'Training')
ref_info = ref_info.rename(columns={'class': 'Class'})

# Get molecular weight from RDKit
from rdkit import Chem
from rdkit.Chem import Descriptors

mw_list = []
formula_list = []
for _, row in ref_info.iterrows():
    mol = Chem.MolFromSmiles(row['smiles'])
    if mol is not None:
        mol = Chem.AddHs(mol)
        mw_list.append(Descriptors.ExactMolWt(mol))
        formula_list.append(Chem.rdMolDescriptors.CalcMolFormula(mol))
    else:
        mw_list.append(None)
        formula_list.append(None)

ref_info['Formula'] = formula_list
ref_info['MW (g/mol)'] = mw_list

# Sort by class then name
class_sort_order = {'CFC': 0, 'HCFC': 1, 'HFC': 2, 'HFO': 3, 'PFC': 4, 'HCFO': 5}
ref_info['class_sort'] = ref_info['Class'].map(class_sort_order)
ref_info = ref_info.sort_values(['class_sort', 'Name']).reset_index(drop=True)

table2 = ref_info[['Class', 'Name', 'Formula', 'MW (g/mol)', 'Split']].copy()
table2['MW (g/mol)'] = table2['MW (g/mol)'].apply(lambda x: f"{x:.1f}" if x else "")

print("\n" + "="*70)
print("TABLE 2: Refrigerant Classification and Train/Validation Assignment")
print("="*70)
print(table2.to_string(index=False))
print(f"\nTotal: {len(table2)} refrigerants")
print(f"Training: {(table2['Split']=='Training').sum()}, Validation: {(table2['Split']=='Validation').sum()}")

table2.to_csv(NOTEBOOK_DIR / "results" / "refrigerant_classification.csv", index=False)
print("\nSaved to results/refrigerant_classification.csv")
```

## 10. Supplementary Information: Per-Refrigerant Viscosity Curves vs Pressure

Generate the pressure-axis curves cited in the Supplementary Information: pressure on the
horizontal axis and one line per sampled temperature. Files are written to
`results/appendix/` with the `app_P_` prefix used by the supplementary manuscript.


```python
# Supplementary figures: viscosity vs pressure, one line per temperature.

(NOTEBOOK_DIR / "results" / "appendix").mkdir(exist_ok=True)

# Combine primary-model predictions and state variables for both splits.
plot_df = results_df.copy()
plot_df["Temperature"] = np.concatenate([train_T, val_T])
plot_df["Pressure"] = np.concatenate([train_P, val_P])

class_sort = {"CFC": 0, "HCFC": 1, "HFC": 2, "HFO": 3, "PFC": 4, "HCFO": 5}
all_refs_sorted = sorted(
    df["Name"].unique(),
    key=lambda name: (class_sort.get(class_map.get(name, ""), 99), name),
)

# Round temperature for grouping (same trick as P_round above).
plot_df['T_round'] = plot_df['Temperature'].round(2)  # K

for ref_name in all_refs_sorted:
    ref_data = plot_df[plot_df['Name'] == ref_name].sort_values('Pressure')
    if ref_data.empty:
        continue

    is_val = ref_name in val_refrigerants
    split_label = 'VALIDATION' if is_val else 'Training'
    cls = class_map.get(ref_name, '?')

    fig, ax = plt.subplots(figsize=(5, 3.5))

    if is_val:
        for spine in ax.spines.values():
            spine.set_edgecolor(COLOR_VAL)
            spine.set_linewidth(2.0)
    else:
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)

    temperatures_unique = sorted(ref_data['T_round'].unique())
    cmap = plt.cm.viridis(np.linspace(0.15, 0.85, len(temperatures_unique)))

    for i, t_val in enumerate(temperatures_unique):
        subset = ref_data[ref_data['T_round'] == t_val].sort_values('Pressure')
        # Pressure in MPa on the x-axis (data are stored in Pa).
        p_mpa = subset['Pressure'].to_numpy() / 1e6
        ax.plot(p_mpa, subset['Actual'], 'o-',
                color=cmap[i], markersize=4, linewidth=1.2,
                label=f'{t_val:.1f} K')
        ax.plot(p_mpa, subset['Predicted'], 's--',
                color=cmap[i], markersize=3, linewidth=1.0, alpha=0.8)

    # Style legend entries (actual vs predicted).
    ax.plot([], [], 'ko-', markersize=4, linewidth=1.2, label='Actual')
    ax.plot([], [], 'ks--', markersize=3, linewidth=1.0, alpha=0.8, label='Predicted')

    ax.set_xlabel(r'Pressure (MPa)')
    ax.set_ylabel(r'Viscosity ($\mu$Pa$\cdot$s)')
    ax.set_yscale('log')
    ax.set_title(f'{ref_name} -- {cls} [{split_label}]', fontsize=11,
                 fontweight='bold' if is_val else 'normal')
    ax.legend(fontsize=7, ncol=2, loc='best')
    ax.grid(True, alpha=0.15, which='both')

    plt.tight_layout()
    plt.savefig(NOTEBOOK_DIR / "results" / "appendix" / f"app_P_{ref_name}.pgf", bbox_inches='tight')
    plt.savefig(NOTEBOOK_DIR / "results" / "appendix" / f"app_P_{ref_name}.pdf", bbox_inches='tight')
    plt.close()

print(f"Generated {len(all_refs_sorted)} supplementary per-refrigerant figures.")
print(f"Saved to results/appendix/app_P_*.{{pgf,pdf}}")
```
